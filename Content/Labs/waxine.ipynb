# Waxinekaarsje

Als je een waxinekaarsje aan steekt en er een glas overheen zet, zal het kaarsje uit gaan. 
Als je dat waxinekaarsje laat drijven op water terwijl je er het glas er overheen zet, zal het water in het glas stijgen.
Op het internet zwerven er verschillende verklaringen voor dit fenomeen rond.
1. Het kaarsje verbruikt zuurstof, waardoor er een vacuüm ontstaat en het water omhoog wordt gezogen.
2. Het kaarsje verwarmt de lucht in het glas, waardoor de lucht uitzet. Wanneer de kaars dooft koelt de lucht af waardoor de druk afneemt en er water omhoog wordt gezogen.
3. Bij het verbranden van de kaars ontstaan er waterdamp en koolstofdioxide. De waterdamp condenseert aan de binnenkant van het glas, waardoor er minder gas in het glas is en de druk afneemt. Hierdoor wordt het water omhoog gezogen.

```{exercise}
Onderzoek welke verklaring het meest waarschijnlijk is en leg uit waarom de andere verklaringen minder waarschijnlijk zijn.
```

Tijdens het experiment steeg het water in het glas ongeveer 1–3 cm nadat de kaars doofde. Voor een glas met een straal van ongeveer 3 cm komt dit overeen met een volumetoename van ongeveer 40-60 mL water. Dit betekent dat het gasvolume in het glas met enkele tientallen milliliters is afgenomen, wat ongeveer 10–25% is. 

Verklaring 1 is onwaarschijnlijk, want lucht bestaat uit ongeveer 21% zuurstof. Zelfs als alle zuurstof volledig zou worden verbruikt, ontstaan er ook weer moleculen bij verbranding. Een volumeverandering van hoogstens enkele procenten is dus onvoldoende om een waterstijging van 10–25% te verklaren. Verklaring 1 is daarom kwantitatief onwaarschijnlijk.

Verklaring 3 beweert dat condensatie van waterdamp de druk verlaagt. wat bij kamertemperatuur gaat dit over ongeveer 2,6 kPa. Wanneer de lucht in het glas afkoelt van bijvoorbeeld 60 °C naar 22 °C, kan een deel van de waterdamp condenseren. Maar ook dit effect alleen is dus te klein om de volumeverandering volledig te verklaren.

Verklaring 2 kan daarintegen wel het waargenomen verschil verklaren. Volgens de ideale gaswet is bij constant volume de druk recht evenredig met de absolute temperatuur, wat betekent dat bij een temperatuurverschil veroorzaakt door het kaarsje de druk tijdelijk met ongeveer 19% kan veranderen. Omdat er tijdens de verwarmingsfase gas is ontsnapt, bevat het glas nu minder moleculen en krijg je een drukverschil van tientallen kilopascal. Dit is wel voldoende om een volumeverschil van 10–20% te veroorzaken, wat overeenkomt met de wat er tijdens het experiment is gezien.
